```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define physical constants (SI units)
k_B = 1.380649e-23  # Boltzmann constant (J/K)
m_e = 9.109383e-31  # Mass of an electron (kg)

# 2. Define plasma parameters
T_eV = 1.0          # Temperature in electron-volts (eV)
T_K = T_eV * 11604  # Convert eV to Kelvin (1 eV ≈ 11,604 K)

# 3. Calculate thermal velocity (v_th)
# This helps us set a reasonable range for the x-axis
v_th = np.sqrt(k_B * T_K / m_e)

# 4. Generate velocity data points (from -3 to +3 thermal velocity)
v = np.linspace(-3 * v_th, 3 * v_th, 1000)

# 5. Define the 1D Maxwellian Distribution Function
def maxwellian_1d(v, m, T):
    coefficient = np.sqrt(m / (2 * np.pi * k_B * T))  # Normalization factor
    exponent = - (m * v**2) / (2 * k_B * T)
    return coefficient * np.exp(exponent)

# Calculate distribution values
f_v = maxwellian_1d(v, m_e, T_K)

# 6. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(v, f_v, label=f'Electron Temp = {T_eV} eV', color='blue', linewidth=2)

# Format the graph
plt.title('1D Maxwellian Velocity Distribution Function for Electrons', fontsize=12, fontweight='bold')
plt.xlabel('Velocity $v$ (m/s)', fontsize=10)
plt.ylabel('Probability Density $f(v)$ (s/m)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)

# Display the plot
plt.tight_layout()
plt.show()
```

### 2D Matrix Manipulation and Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

# Define 1D vectors for x and y coordinates
x_coords = np.array([1, 2, 3, 4])
y_coords = np.array([10, 20, 30])

print("Original x_coords:", x_coords)
print("Original y_coords:", y_coords)

# Use np.meshgrid to create 2D coordinate matrices
X, Y = np.meshgrid(x_coords, y_coords)

print("\nX matrix (x-coordinates replicated across rows):\n", X)
print("\nY matrix (y-coordinates replicated across columns):\n", Y)

# Visualize the meshgrid output as a scatter plot to show the grid points
plt.figure(figsize=(7, 6))
plt.scatter(X, Y, color='blue', s=100, marker='o', edgecolors='black')

# Annotate points for clarity
for i in range(Y.shape[0]):
    for j in range(X.shape[1]):
        plt.text(X[i, j] + 0.1, Y[i, j] + 0.5, f'({X[i, j]}, {Y[i, j]})', fontsize=9, ha='left', va='bottom')

plt.title('Visualization of np.meshgrid Output', fontsize=14, fontweight='bold')
plt.xlabel('X-coordinate', fontsize=12)
plt.ylabel('Y-coordinate', fontsize=12)
plt.xticks(x_coords)
plt.yticks(y_coords)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range for x and y
x = np.linspace(-5, 5, 100) # 100 points between -5 and 5
y = np.linspace(-5, 5, 100) # 100 points between -5 and 5

# Create the coordinate surface matrix using np.meshgrid
X, Y = np.meshgrid(x, y)

# Define a sample 2D function (e.g., a paraboloid)
Z = X**2 + Y**2

# Create the 2D plot (e.g., a contour plot)
plt.figure(figsize=(8, 6))
plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(label='Z value')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Graph of Z = X^2 + Y^2')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Create the simple 1D input vectors (the raw "threads")
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 30, 40])

# 2. Use meshgrid to weave them into a 2D coordinate surface matrix
X, Y = np.meshgrid(x, y)

# 3. Set up the plot
plt.figure(figsize=(8, 6))

# Plot the horizontal grid lines (Rows replicated from x)
for row in Y:
    plt.axhline(y=row[0], color="lightblue", linestyle="--", alpha=0.7)

# Plot the vertical grid lines (Columns replicated from y)
for col in X[0]:
    plt.axvline(x=col, color="lightcoral", linestyle="--", alpha=0.7)

# 4. Plot the intersecting points (the woven matrix)
# X and Y are 2D arrays, so we flatten them to scatter individual points
plt.scatter(
    X.flatten(),
    Y.flatten(),
    color="purple",
    s=100,
    zorder=3,
    label="Meshgrid Intersections (X, Y)",
)

# 5. Add text labels to show the coordinates of each point
for i in range(len(y)):
    for j in range(len(x)):
        plt.text(
            X[i, j],
            Y[i, j] + 1,
            f"({X[i, j]}, {Y[i, j]})",
            ha="center",
            fontsize=9,
        )

# Customize the chart appearance
plt.title("How np.meshgrid(x, y) Weaves a 2D Grid", fontsize=14, pad=15)
plt.xlabel("X vector values (Horizontal)", fontsize=12)
plt.ylabel("Y vector values (Vertical)", fontsize=12)
plt.xticks(x)
plt.yticks(y)
plt.xlim(0, 6)
plt.ylim(5, 45)
plt.grid(False)  # Turn off default grid to see our custom "woven" lines clearly
plt.legend(loc="upper left")

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Set up the data space
# Generates 500 points between -1.5 and 1.5 for the Velocity Grid Space
x = np.linspace(-1.5, 1.5, 500)

# 2. Create the "Theoretical Theory" curve
# We use a sine wave that dampens slightly, multiplying by 1e-6 to match your scale
y_theoretical = np.sin(2 * np.pi * x) * np.exp(-0.2 * x**2) * 1e-6

# 3. Create the "Noisy Satellite Observations"
# We add random Gaussian noise to the theoretical data
np.random.seed(42)  # Keeps the noise exactly the same every time you run it
noise = np.random.normal(0, 0.25 * 1e-6, size=x.shape)
y_noisy = y_theoretical + noise

# 4. Plotting the data
plt.figure(figsize=(9, 6))

# Plot the noisy data as a light scatter plot or thin line
plt.plot(x, y_noisy, color='#9ecae1', alpha=0.7, label='Noisy Satellite Observations', linewidth=1.5)

# Plot the theoretical data as a clean, bold line
plt.plot(x, y_theoretical, color='#08519c', label='Theoretical Theory', linewidth=2.5)

# 5. Formatting axes and labels
plt.title('Computational Reality: Perfect Theory vs. Noisy Observations', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Velocity Grid Space (m/s)', fontsize=10)
plt.ylabel('Measured Value ($10^{-6}$)', fontsize=10)

# Adjusting the axes limits to match your description
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5e-6, 1.5e-6)

# Formatting the ticks to show clean decimals (the 10^-6 multiplier is in the axis label)
plt.xticks(np.arange(-1.5, 1.6, 0.5))
plt.yticks(np.arange(-1.5e-6, 1.6e-6, 0.5e-6), ['-1.5', '-1.0', '-0.5', '0.0', '0.5', '1.0', '1.5'])

# Adding a scientific multiplier notation text box if desired (like the "1e-6" or "10^-6" on charts)
plt.text(1.52, 1.5e-6, '1e-6', fontsize=9, va='bottom')

# 6. Styling the chart
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='none')
plt.tight_layout()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Set up the grid size
grid_size = 500
x = np.linspace(-10, 10, grid_size)
y = np.linspace(-10, 10, grid_size)

# Create the 2D coordinate grid (Matrix Geometry)
X, Y = np.meshgrid(x, y)

# 2. Define the matrix geometry parameters
x_offset = 0.0  # Shifts structure left/right
y_offset = 0.0  # Shifts structure up/down
width_x = 4.0  # Controls horizontal spread
width_y = 4.0  # Controls vertical spread (keep equal to width_x for a circle)

# 3. Apply the Multi-Dimensional Equation
# This calculates the distance value for every single point in the matrix
Z = ((X - x_offset) / width_x) ** 2 + ((Y - y_offset) / width_y) ** 2

# 4. Plot the colorful graph
plt.figure(figsize=(7, 7))

# 'imshow' displays the matrix as an image.
# 'cmap' sets the color palette. 'jet', 'plasma', or 'viridis' look very colorful!
plt.imshow(
    Z,
    extent=[x.min(), x.max(), y.min(), y.max()],
    origin="lower",
    cmap="jet",
)

# Add a color bar to show the value scale
plt.colorbar(label="Matrix Equation Value")

# Add labels and title
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Colorful 2D Matrix Geometry (Perfect Circle)")

# Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Set up the grid size
grid_size = 500
x = np.linspace(-10, 10, grid_size)
y = np.linspace(-10, 10, grid_size)

# Create the 2D coordinate grid (Matrix Geometry)
X, Y = np.meshgrid(x, y)

# 2. Define the matrix geometry parameters for an OVAL
x_offset = 0.0  # Shifts structure left/right
y_offset = 0.0  # Shifts structure up/down
width_x = 6.0  # Wider horizontal spread
width_y = 3.0  # Narrower vertical spread (making it an oval!)

# 3. Apply the Multi-Dimensional Equation
# This calculates the distance value for every single point in the matrix
Z = ((X - x_offset) / width_x) ** 2 + ((Y - y_offset) / width_y) ** 2

# 4. Plot the colorful graph in a normal screen size (6x4 inches)
plt.figure(figsize=(6, 4))

# 'imshow' displays the matrix as an image.
# 'cmap="plasma"' gives a beautiful, vibrant, colorful look!
plt.imshow(
    Z,
    extent=[x.min(), x.max(), y.min(), y.max()],
    origin="lower",
    cmap="plasma",
    aspect="auto",
)

# Add a color bar to show the value scale
plt.colorbar(label="Matrix Equation Value")

# Add labels and title
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Colorful 2D Matrix Geometry (Oval Shape)")

# Display the graph
plt.show()
```
